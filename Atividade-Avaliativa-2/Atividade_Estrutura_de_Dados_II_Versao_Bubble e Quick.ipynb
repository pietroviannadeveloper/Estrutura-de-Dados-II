# 1. Bubble Sort

## O que o codigo faz
O Bubble Sort compara elementos adjacentes e realiza trocas quando eles estao fora de ordem.

## Objetivo
Contabilizar comparacoes e trocas e observar o comportamento do algoritmo para diferentes tamanhos de entrada.

## Complexidade
- Melhor caso: **O(n)** com parada antecipada
- Caso medio: **O(n²)**
- Pior caso: **O(n²)**


```python
def bubble_sort(arr):
    a = arr[:]
    n = len(a)
    comparisons = swaps = 0

    for i in range(n - 1):
        trocou = False
        for j in range(n - 1 - i):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                trocou = True
        if not trocou:
            break

    return a, comparisons, swaps
```

### Teste rapido
O vetor abaixo serve apenas para conferir se a funcao esta ordenando corretamente.

```python
vetor_teste = [5, 1, 4, 2, 8]
ordenado, comparacoes, trocas = bubble_sort(vetor_teste)
print('Original:', vetor_teste)
print('Ordenado:', ordenado)
print('Comparacoes:', comparacoes)
print('Trocas:', trocas)
```

# 2. Quick Sort

## O que o codigo faz
O Quick Sort divide o vetor em particoes a partir de um pivo e repete o processo recursivamente. Nesta implementacao, o ultimo elemento e usado como pivo.

## Objetivo
Contabilizar comparacoes e movimentacoes para comparar seu crescimento com o Bubble Sort.

## Complexidade
- Melhor caso: **O(n log n)**
- Caso medio: **O(n log n)**
- Pior caso: **O(n²)**


```python
def quick_sort(arr):
    a = arr[:]
    comparisons = movements = 0

    def partition(lo, hi):
        nonlocal comparisons, movements
        pivot = a[hi]
        i = lo - 1
        for j in range(lo, hi):
            comparisons += 1
            if a[j] <= pivot:
                i += 1
                a[i], a[j] = a[j], a[i]
                movements += 1
        a[i + 1], a[hi] = a[hi], a[i + 1]
        movements += 1
        return i + 1

    def qs(lo, hi):
        if lo < hi:
            p = partition(lo, hi)
            qs(lo, p - 1)
            qs(p + 1, hi)

    qs(0, len(a) - 1)
    return a, comparisons, movements
```

### Teste rapido

```python
vetor_teste = [5, 1, 4, 2, 8]
ordenado, comparacoes, movimentacoes = quick_sort(vetor_teste)
print('Original:', vetor_teste)
print('Ordenado:', ordenado)
print('Comparacoes:', comparacoes)
print('Movimentacoes:', movimentacoes)
```

# 3. Comparacao experimental

Os dois algoritmos recebem copias do mesmo vetor. Foi usada a semente `42` para que os dados possam ser reproduzidos em outra execucao.

```python
import random

random.seed(42)

for n in [10, 20, 1000]:
    original = [random.randint(0, 100000) for _ in range(n)]

    copia_bubble = original[:]
    copia_quick = original[:]

    _, b_cmp, b_swap = bubble_sort(copia_bubble)
    _, q_cmp, q_mov = quick_sort(copia_quick)

    print(n, b_cmp, b_swap, q_cmp, q_mov)
```

### O que observar
Para 10 e 20 elementos a diferenca existe, mas em 1.000 elementos ela se torna muito mais evidente. Isso e coerente com O(n²) para o Bubble Sort e O(n log n) para o Quick Sort no caso medio.

# 4. Busca sequencial em matriz

## Objetivo
Percorrer uma matriz com loops aninhados, localizar um valor e registrar linha, coluna e quantidade de comparacoes.

No pior caso, a complexidade e **O(m x n)**.

```python
def busca_sequencial_matriz(matriz, valor):
    comparacoes = 0

    for i in range(len(matriz)):
        for j in range(len(matriz[0])):
            comparacoes += 1
            if matriz[i][j] == valor:
                return True, i, j, comparacoes

    return False, -1, -1, comparacoes
```

# 5. Testes de busca em matrizes

Serão avaliados tres cenarios: valor no inicio, valor na ultima posicao e valor inexistente.

```python
def criar_matriz_sequencial(linhas, colunas):
    return [[i * colunas + j for j in range(colunas)] for i in range(linhas)]

for tamanho in [2, 10, 100]:
    matriz = criar_matriz_sequencial(tamanho, tamanho)
    inicio = busca_sequencial_matriz(matriz, matriz[0][0])
    fim = busca_sequencial_matriz(matriz, matriz[-1][-1])
    inexistente = busca_sequencial_matriz(matriz, -1)

    print(f'\nMatriz {tamanho} x {tamanho}')
    print('Total:', tamanho * tamanho)
    print('Inicio:', inicio[3], 'comparacoes')
    print('Final:', fim[3], 'comparacoes')
    print('Inexistente:', inexistente[3], 'comparacoes')
```

# 6. Hands On 1 - Array de temperaturas

## Objetivo
Calcular media, maior e menor temperatura, indices correspondentes e quantidade de valores acima da media.

```python
temperaturas = [19.0, 24.3, 21.3, 25.3, 25.6,
                16.1, 15.2, 29.2, 19.4, 19.0]

soma = 0.0
for t in temperaturas:
    soma += t
media = soma / len(temperaturas)

maior = menor = temperaturas[0]
idx_maior = idx_menor = 0
for i in range(1, len(temperaturas)):
    if temperaturas[i] > maior:
        maior, idx_maior = temperaturas[i], i
    if temperaturas[i] < menor:
        menor, idx_menor = temperaturas[i], i

acima_media = sum(1 for t in temperaturas if t > media)

print(f'Media: {media:.2f} °C')
print(f'Maior: {maior:.1f} °C - indice {idx_maior}')
print(f'Menor: {menor:.1f} °C - indice {idx_menor}')
print('Acima da media:', acima_media)
print('Complexidade: O(n)')
```
